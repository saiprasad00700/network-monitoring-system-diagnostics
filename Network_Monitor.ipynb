```python
import socket

def dns_lookup(host):
    ip = socket.gethostbyname(host)
    return ip

print("IP for google:", dns_lookup("google.com"))
print("IP for github:", dns_lookup("github.com"))
```

```python
import subprocess
import platform

def ping_host(host):
    if dns_lookup(host) is None:
        return False

    param = "-n" if platform.system().lower() == "windows" else "-c"

    command = ["ping", param, "1", host]

    result = subprocess.run(command, capture_output=True, text=True)

    return result.returncode == 0

print(ping_host("google.com"))
print(ping_host("abcxyz123.com"))
```

```python
import socket

def scan_port(host, port):
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.settimeout(1)

    result = s.connect_ex((host, port))
    s.close()

    return result == 0
print("Port 22:", scan_port("google.com", 22))
print("Port 80:", scan_port("google.com", 80))
print("Port 443:", scan_port("google.com", 443))
```

```python
ports = [22, 80, 443]

for port in ports:
    if scan_port("google.com", port):
        print(f"Port {port} is OPEN")
    else:
        print(f"Port {port} is CLOSED")
```

```python
def network_diagnostics(host):

    print("=" * 50)
    print("   NETWORK MONITORING & SYSTEM DIAGNOSTICS")
    print("=" * 50)

    print("Time :", datetime.now().strftime("%d-%m-%Y %H:%M:%S"))
    print("Host :", host)

    # DNS Lookup
    ip = dns_lookup(host)
    print("IP   :", ip)

    # Ping Status
    status = "Reachable" if ping_host(host) else "Unreachable"
    print("Status :", status)

    print("\nPort Scan")
    print("-" * 50)

    services = {
        22: "SSH",
        80: "HTTP",
        443: "HTTPS"
    }

    open_ports = 0

    for port, service in services.items():

        if scan_port(host, port):
            state = "OPEN"
            open_ports += 1
        else:
            state = "CLOSED"

        print(f"{port:<5} {service:<10} {state}")

    print("-" * 50)
    print(f"Total Open Ports : {open_ports}")
    print("=" * 50)
```

```python
host = input("Enter domain: ")

network_diagnostics(host)
```
